```python
import pandas as pd
import numpy as np

# Reproducible experiment
SEED = 42
np.random.seed(SEED)

# Get speakers by gender
female_speakers = (
    gender_df[gender_df["label"] == "female"]["speaker_id"]
    .unique()
)

male_speakers = (
    gender_df[gender_df["label"] == "male"]["speaker_id"]
    .unique()
)

# We have 21 female speakers.
# Select the same number of male speakers.
N_SPEAKERS = min(len(female_speakers), len(male_speakers))

rng = np.random.default_rng(SEED)

selected_female = rng.choice(
    female_speakers,
    size=N_SPEAKERS,
    replace=False
)

selected_male = rng.choice(
    male_speakers,
    size=N_SPEAKERS,
    replace=False
)

selected_speakers = np.concatenate([
    selected_female,
    selected_male
])

print("Selected female speakers:", len(selected_female))
print("Selected male speakers:", len(selected_male))
print("Total selected speakers:", len(selected_speakers))
```

```python
from sklearn.model_selection import train_test_split

# Split speakers, NOT recordings
train_speakers, temp_speakers = train_test_split(
    selected_speakers,
    test_size=0.30,
    random_state=SEED,
    stratify=[
        "female" if s in selected_female else "male"
        for s in selected_speakers
    ]
)

val_speakers, test_speakers = train_test_split(
    temp_speakers,
    test_size=0.50,
    random_state=SEED,
    stratify=[
        "female" if s in selected_female else "male"
        for s in temp_speakers
    ]
)

print("TRAIN speakers:", len(train_speakers))
print("VALIDATION speakers:", len(val_speakers))
print("TEST speakers:", len(test_speakers))
```

```python
MAX_CLIPS_PER_SPEAKER = 50

def make_split_df(df, speakers, max_clips=50, seed=42):
    split_df = df[df["speaker_id"].isin(speakers)].copy()

    # Limit recordings from each speaker
    split_df = (
        split_df
        .groupby("speaker_id", group_keys=False)
        .apply(
            lambda x: x.sample(
                n=min(len(x), max_clips),
                random_state=seed
            )
        )
        .reset_index(drop=True)
    )

    return split_df


train_df = make_split_df(gender_df, train_speakers, MAX_CLIPS_PER_SPEAKER)
val_df   = make_split_df(gender_df, val_speakers, MAX_CLIPS_PER_SPEAKER)
test_df  = make_split_df(gender_df, test_speakers, MAX_CLIPS_PER_SPEAKER)

print("Training recordings   :", len(train_df))
print("Validation recordings :", len(val_df))
print("Test recordings       :", len(test_df))
```

```python
import librosa
import numpy as np

SAMPLE_RATE = 16000
DURATION = 4
NUM_SAMPLES = SAMPLE_RATE * DURATION

N_MELS = 64
N_FFT = 1024
HOP_LENGTH = 256
FMIN = 50
FMAX = 8000


def load_audio_features(path, training=False):
    # Load as mono, 16 kHz
    audio, sr = librosa.load(
        path,
        sr=SAMPLE_RATE,
        mono=True
    )

    audio = audio.astype(np.float32)

    # Normalize
    max_val = np.max(np.abs(audio))

    if max_val > 0:
        audio = audio / max_val

    # ---- Fixed 4-second length ----

    if len(audio) > NUM_SAMPLES:

        if training:
            # Random crop during training
            start = np.random.randint(
                0,
                len(audio) - NUM_SAMPLES + 1
            )
        else:
            # Center crop for validation/test
            start = (len(audio) - NUM_SAMPLES) // 2

        audio = audio[start:start + NUM_SAMPLES]

    else:
        # Zero padding
        padding = NUM_SAMPLES - len(audio)

        audio = np.pad(
            audio,
            (0, padding),
            mode="constant"
        )

    # ---- Mel spectrogram ----

    mel = librosa.feature.melspectrogram(
        y=audio,
        sr=SAMPLE_RATE,
        n_mels=N_MELS,
        n_fft=N_FFT,
        hop_length=HOP_LENGTH,
        fmin=FMIN,
        fmax=FMAX
    )

    # Convert to dB
    log_mel = librosa.power_to_db(
        mel,
        ref=np.max
    )

    # Normalize features approximately to 0–1
    log_mel = (log_mel + 80.0) / 80.0

    return log_mel.astype(np.float32)
```

```python
import torch
from torch.utils.data import Dataset, DataLoader


LABEL_MAP = {
    "male": 0,
    "female": 1
}


class GenderDataset(Dataset):

    def __init__(self, dataframe, training=False):
        self.df = dataframe.reset_index(drop=True)
        self.training = training

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        # Extract Mel features
        features = load_audio_features(
            row["audio_path"],
            training=self.training
        )

        # CNN expects:
        # (channels, mel, time)
        features = torch.tensor(
            features,
            dtype=torch.float32
        ).unsqueeze(0)

        label = torch.tensor(
            LABEL_MAP[row["label"]],
            dtype=torch.long
        )

        return features, label
```

```python
import torch
import torch.nn as nn


class GenderCNN(nn.Module):

    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(

            # 1 → 16
            nn.Conv2d(1, 16, kernel_size=3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.MaxPool2d(2),

            # 16 → 32
            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2),

            # 32 → 64
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2),

            # Regularization
            nn.Dropout(0.30)
        )

        # Instead of calculating the flattened size manually,
        # use Adaptive Average Pooling.
        self.pool = nn.AdaptiveAvgPool2d((1, 1))

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.30),
            nn.Linear(32, 2)
        )

    def forward(self, x):

        x = self.features(x)
        x = self.pool(x)
        x = self.classifier(x)

        return x


model = GenderCNN()

print(model)
```

```python
import torch
import torch.nn as nn

male_count = (train_df["label"] == "male").sum()
female_count = (train_df["label"] == "female").sum()

total = male_count + female_count

male_weight = total / (2 * male_count)
female_weight = total / (2 * female_count)

class_weights = torch.tensor(
    [male_weight, female_weight],
    dtype=torch.float32
)

print("Male recordings  :", male_count)
print("Female recordings:", female_count)

print("Male weight      :", male_weight)
print("Female weight    :", female_weight)
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model = GenderCNN().to(device)

criterion = nn.CrossEntropyLoss(
    weight=class_weights.to(device)
)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.001,
    weight_decay=1e-4
)

print("Device:", device)
```

```python
def train_one_epoch(model, loader, criterion, optimizer, device):

    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for features, labels in loader:

        features = features.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(features)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        running_loss += loss.item() * features.size(0)

        predictions = outputs.argmax(dim=1)

        correct += (
            predictions == labels
        ).sum().item()

        total += labels.size(0)

    loss = running_loss / total
    accuracy = correct / total

    return loss, accuracy
```

```python
def evaluate(model, loader, criterion, device):

    model.eval()

    running_loss = 0.0
    correct = 0
    total = 0

    all_predictions = []
    all_labels = []

    with torch.no_grad():

        for features, labels in loader:

            features = features.to(device)
            labels = labels.to(device)

            outputs = model(features)

            loss = criterion(outputs, labels)

            running_loss += loss.item() * features.size(0)

            predictions = outputs.argmax(dim=1)

            correct += (
                predictions == labels
            ).sum().item()

            total += labels.size(0)

            all_predictions.extend(
                predictions.cpu().numpy()
            )

            all_labels.extend(
                labels.cpu().numpy()
            )

    loss = running_loss / total
    accuracy = correct / total

    return loss, accuracy, all_labels, all_predictions
```

```python
EPOCHS = 20

best_val_accuracy = 0.0
best_model_path = "/content/best_gender_cnn.pt"

for epoch in range(EPOCHS):

    train_loss, train_acc = train_one_epoch(
        model,
        train_loader,
        criterion,
        optimizer,
        device
    )

    val_loss, val_acc, _, _ = evaluate(
        model,
        val_loader,
        criterion,
        device
    )

    print(
        f"Epoch {epoch+1:02d}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.3f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.3f}"
    )

    if val_acc > best_val_accuracy:

        best_val_accuracy = val_acc

        torch.save(
            model.state_dict(),
            best_model_path
        )

print("\nBest validation accuracy:",
      best_val_accuracy)

print("Best model saved to:",
      best_model_path)
```

```python
import torch
import librosa
import numpy as np


def predict_gender(audio_path):

    # Extract exactly the same features used during training
    features = load_audio_features(
        audio_path,
        training=False
    )

    # [64, 251] → [1, 1, 64, 251]
    tensor = torch.tensor(
        features,
        dtype=torch.float32
    ).unsqueeze(0).unsqueeze(0)

    tensor = tensor.to(device)

    model.eval()

    with torch.no_grad():

        output = model(tensor)

        probabilities = torch.softmax(
            output,
            dim=1
        )[0]

    male_prob = probabilities[0].item()
    female_prob = probabilities[1].item()

    predicted_class = torch.argmax(probabilities).item()

    label = {
        0: "male",
        1: "female"
    }[predicted_class]

    confidence = probabilities[predicted_class].item()

    return label, confidence, male_prob, female_prob
```
